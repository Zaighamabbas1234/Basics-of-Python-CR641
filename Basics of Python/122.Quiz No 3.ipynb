```python
# Write a program that cheaks is a number is even or odd?
```

```python
# Solution:-
```

```python
number = int(input("Please enter the number?"))
```

```python
print(number)
```

```python
if (number % 2 == 0):
    print("This is a even number.")
else:
    print("This is a odd number.")
```
