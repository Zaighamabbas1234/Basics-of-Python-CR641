```python
# Write a python program that finds the largest numbers in a list using a for loop?
```

```python
# Solution:-
```

```python
numbers = [10,20,30,40,50]
```

```python
max_number = numbers[0]
```

```python
for number in numbers:
    if number > max_number:
        max_number = number
        print("largest number is", max_number)
```
