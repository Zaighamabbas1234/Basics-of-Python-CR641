```python
# Write a python program that asks the user to guess a number between 1 and 10. Use a while looop to keep asking until the user gusses the correct number (5)?
```

```python
# Solution:-
```

```python
number = 5
```

```python
while True:
    guess = int(input("Enter the number between 1 and 10."))
    if guess == number:
        print("Congratulations! you gussed it right.")
    else:
        print("Try Again!")
```
