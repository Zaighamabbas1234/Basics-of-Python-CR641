```python
# Write a program that calculates the sum of the numbers from 1 to n using the while loop?
```

```python
# Solution:-
```

```python
number = int(input("Please enter the number?"))
```

```python
print(number)
```

```python
i = 1
```

```python
sum = 0
```

```python
while (i<=number):
    sum += i
    i += 1
    print(sum)
```
