```python
# Write a python program that counts the number of vowels in a  given string using a for loop?
```

```python
# Solution:-
```

```python
s = input("Enter a string.").lower()
vowels = "aeiou"
count = 0
for char in s:
    if char in vowels:
        count += 1
        print("Numbers of vowels:", count)
```
