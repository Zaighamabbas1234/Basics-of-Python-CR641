```python
grade = str(input("Enter the grade (A,B,C,D,E)."))
```

```python
print(grade)
```

```python
if (grade.upper() == "A"):
    print("Excellent.")
elif (grade.upper() == "B"):
    print("Good.")
elif (grade.upper() =="C"):
    print("Fair.")
elif (grade.upper() == "D"):
    print("Not Bad.")
elif (grade.upper() == "E"):
    print("Bad.")
else:
    print("Fail.")
```
